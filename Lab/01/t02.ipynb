```python
from typing import Dict, TypedDict, List, Literal

type StudentMajor = Literal["AI", "CS", "SE"]

class Student(TypedDict):
    Name: str
    Major: StudentMajor
    Grades: List[Literal["A", "B", "C", "D", "E", "F"]]

type StudentMap = Dict[str, Student]

def get_highest_avg_std_name(stds: StudentMap) -> str:
    def _key(std: Student):
        grades = std["Grades"]
        return sum(map(lambda x: ord(x) - ord("F"), grades)) / len(grades)
    return stds[max(stds, key=lambda x: _key(stds[x]))]["Name"]

def get_major_stds(stds: StudentMap, mjr: StudentMajor) -> List[Student]:
    return [std for std in stds.values() if std["Major"] == mjr]

students: StudentMap = {
    "1": {
        "Name": "Test 1",
        "Major": "AI",
        "Grades": ["A", "A", "B", "C"]
    },
    "2": {
        "Name": "Test 2",
        "Major": "AI",
        "Grades": ["A", "A", "A", "E"]
    },
    "3": {
        "Name": "Test 3",
        "Major": "CS",
        "Grades": ["A", "A", "A", "A", "F", "E"]
    }
}

print("Name:", get_highest_avg_std_name(students))
print("Students:", __import__("json").dumps(get_major_stds(students, "AI"), indent=2))
```
