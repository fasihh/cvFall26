```python
from typing import Dict, Tuple

class GroceryManager:
    def __init__(self):
        self.items: Dict[str, Tuple[int, float]] = {}
    
    def add_item(self, item: str, qty: int, price: float):
        self.items[item] = (qty, price)
    
    def remove_item(self, item: str):
        try:
            self.items.pop(item)
        except KeyError:
            raise RuntimeError(f"{item} does not exist in item list")

    def calculate_amount(self) -> float:
        return sum(map(lambda x: x[0] * x[1], self.items.values()))
```

```python
gm = GroceryManager()

gm.add_item("test1", 1, 2.4)
gm.add_item("test2", 4, 2.6)
gm.add_item("test3", 1, 2.4)

gm.remove_item("test1")

gm.calculate_amount()
```
