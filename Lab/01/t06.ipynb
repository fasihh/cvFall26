```python
import os
import cv2
import matplotlib.pyplot as plt
```

```python
image_path = 'Sukuna.jpg'
if not os.path.exists(image_path):
    raise FileNotFoundError(f"{image_path} does not exist")
img = cv2.imread(image_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img.shape
```

```python
N, M = img.shape[:2]
overlay = img.copy()

band_h = int(N * 0.2)
cv2.rectangle(overlay, (0, N - band_h), (M, N), (255, 0, 0), thickness=-1)

blended = cv2.addWeighted(overlay, 0.5, img, 0.5, 0)
```

```python
cv2.putText(
    blended,
    "King of Curses",
    (10, N - band_h // 2),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.7,
    (255, 255, 255),
    2
)
```

```python
plt.figure(figsize=(8, 6))
plt.axis('off')

plt.imshow(blended)
plt.title("Alpha Blended Caption", fontsize=16, color='darkred', pad=15)
plt.show()
```
