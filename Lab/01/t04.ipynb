```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
N, M = 800, 800
img = np.ones((N, M, 3)) * 255

center_N, center_M = N // 2, M // 2

rad, lim = 70, 5
for idx in range(1, lim + 1):
    cv2.circle(img, (center_N, center_M), idx * rad, color=idx % 2, thickness=4)

MAX_RAD = lim * rad
p = np.array((center_N, center_M))
cv2.rectangle(
    img,
    p - MAX_RAD,
    p + MAX_RAD,
    color=0,
    thickness=4
)

plt.imshow(img)
plt.axis('off')

plt.title(f"Center: ({center_N}, {center_M})")

plt.show()
```
